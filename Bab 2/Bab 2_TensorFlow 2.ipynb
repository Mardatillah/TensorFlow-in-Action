Bab 2 membahas TensorFlow 2 secara mendalam, yang merupakan versi terbaru dari framework TensorFlow. TensorFlow 2 dirancang untuk lebih ramah pengembang dengan menyederhanakan banyak aspek dibandingkan versi sebelumnya. Fokus utama TensorFlow 2 adalah pada eager execution, yang memungkinkan pengembang menjalankan perhitungan secara langsung tanpa mendefinisikan graf komputasi terlebih dahulu. Selain itu, TensorFlow 2 mengintegrasikan Keras sebagai submodule utama, mempermudah pembuatan dan pelatihan model pembelajaran mesin.

1. Langkah Pertama dengan TensorFlow 2

    Di bagian ini, pembaca diperkenalkan dengan langkah pertama untuk menginstal dan memulai penggunaan TensorFlow 2. Berikut beberapa aspek yang dibahas:
            
    *   **Instalasi TensorFlow:** Cara menginstal TensorFlow 2 menggunakan pip dan memastikan bahwa TensorFlow berfungsi dengan benar.
    *   **Pengenalan Eager Execution:** TensorFlow 2 menggunakan **eager execution** secara default, yang memungkinkan Anda untuk langsung menjalankan operasi tanpa memerlukan graf komputasi yang sebelumnya harus didefinisikan dalam TensorFlow 1. Ini membuat debugging lebih mudah karena hasil dari operasi langsung terlihat.

2. TensorFlow Building Blocks

    Bab ini menjelaskan tiga komponen dasar dari TensorFlow yang diperlukan untuk membangun model:
    *   **tf.Variable:** Objek ini digunakan untuk menyimpan nilai yang dapat diubah (misalnya, bobot dalam jaringan neural). Variabel ini sangat penting dalam pelatihan model karena nilainya diperbarui selama proses pelatihan.
    *   **tf.Tensor:** Merupakan objek yang menyimpan data dalam bentuk multi-dimensional arrays (tensors). Tensor ini digunakan untuk menyimpan input, output, dan data lainnya yang terlibat dalam perhitungan model.
    *   **tf.Operation:** Ini adalah operasi yang dilakukan pada tensor. Misalnya, perkalian matriks atau fungsi aktivasi yang digunakan dalam jaringan neural.

3. Operasi yang Berhubungan dengan Jaringan Saraf dalam TensorFlow

    Di bagian ini, dibahas beberapa operasi dasar yang sering digunakan dalam membangun model jaringan saraf:
    *   **Perkalian Matriks (Matrix Multiplication):** Dalam jaringan saraf, perkalian matriks digunakan untuk menghitung produk antara bobot dan input. Ini adalah inti dari operasi dalam model deep learning.
    *   **Operasi Konvolusi (Convolution Operation):** Operasi ini digunakan dalam jaringan saraf konvolusional (CNN) untuk mendeteksi pola dalam data spasial seperti gambar. Operasi konvolusi diterapkan untuk memindai fitur dari gambar.
    *   **Operasi Pooling (Pooling Operation):** Pooling digunakan untuk mengurangi dimensi data dan mengurangi kompleksitas perhitungan dalam model CNN. Pooling berfungsi untuk mengumpulkan informasi penting dari data yang lebih besar dan mengurangi ukuran input tanpa kehilangan fitur utama.

Di sini, dibahas berbagai jenis model yang bisa dibangun menggunakan Keras seperti **Sequential**, **Functional**, dan **Sub-classing API**.
    *   **Pengambilan Data untuk Model TensorFlow:** Bagian ini menjelaskan bagaimana mengambil dan memproses data yang akan digunakan untuk melatih model. TensorFlow menyediakan API tf.data untuk membuat pipeline data yang efisien, serta **Keras DataGenerators** untuk mengambil data dari file gambar atau teks. **tensorflow-datasets** juga diperkenalkan sebagai paket untuk mengambil data umum dalam pembelajaran mesin.
