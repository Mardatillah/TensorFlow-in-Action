Bab 8 berfokus pada **segmentasi gambar**, yaitu proses membagi gambar menjadi beberapa bagian atau segmen yang lebih kecil dan bermakna untuk analisis lebih lanjut. **Segmentasi gambar** adalah tugas yang lebih kompleks daripada **klasifikasi gambar**, karena bukan hanya menentukan kategori gambar, tetapi juga memisahkan objek-objek dalam gambar dan memberikan label pada setiap bagian dari gambar.

Bab ini mengajarkan cara menggunakan **Convolutional Neural Networks (CNNs)** untuk melakukan segmentasi gambar, serta berbagai teknik yang dapat digunakan untuk meningkatkan akurasi dan efisiensi model segmentasi. Segmentasi gambar sangat penting dalam berbagai aplikasi, seperti **deteksi objek**, **pengenalan wajah**, **analisis medis**, dan banyak lagi.

### 1. Pengertian dan Pentingnya Segmentasi Gambar
Segmentasi gambar adalah proses yang membagi gambar menjadi beberapa bagian yang lebih kecil untuk memudahkan analisis. Setiap bagian yang dihasilkan dari segmentasi disebut sebagai **segmen**. Setiap segmen tersebut berisi informasi tentang **objek** atau bagian dari objek yang ada dalam gambar.
*   **Tujuan segmentasi** adalah untuk mempermudah pemahaman gambar dengan memisahkan area yang berbeda dalam gambar, seperti objek, latar belakang, atau elemen lainnya.
*   Segmentasi gambar digunakan dalam aplikasi seperti **deteksi penyakit dalam gambar medis**, **analisis citra satelit**, dan **pengenalan objek**.

### 2. Perbedaan antara Segmentasi dan Klasifikasi Gambar
Bab ini membedakan antara **segmentasi gambar** dan **klasifikasi gambar**. Pada klasifikasi gambar, model hanya memberikan satu label untuk keseluruhan gambar (misalnya, gambar tersebut berisi kucing atau anjing). Sedangkan pada segmentasi, model harus memberi label pada setiap piksel dalam gambar untuk menentukan objek apa yang ada di setiap area.
*   **Klasifikasi** memberikan satu label untuk gambar keseluruhan.
*   **Segmentasi** memberikan label untuk setiap piksel dalam gambar, yang memungkinkan model untuk mengenali bagian-bagian objek yang lebih rinci.

### 3. Mask R-CNN untuk Segmentasi Instance
Salah satu teknik yang diperkenalkan dalam bab ini adalah **Mask R-CNN**, yang digunakan untuk melakukan **instance segmentation**, yaitu segmentasi di mana model tidak hanya memisahkan objek, tetapi juga memberikan masker (atau label) untuk setiap objek yang ada dalam gambar.
*   **Mask R-CNN** adalah pengembangan dari **Faster R-CNN** yang sudah sangat sukses dalam deteksi objek. Dengan menambahkan lapisan masker ke Faster R-CNN, Mask R-CNN dapat menghasilkan mask untuk objek yang terdeteksi, memungkinkan segmentasi objek secara tepat.
*   **Komponen Mask R-CNN:**
    *   **Region Proposal Network (RPN)** untuk menghasilkan kandidat objek.
    *   **RoIAlign** untuk memastikan pemetaan posisi yang tepat pada objek.
    *   **Mask Layer** untuk menghasilkan masker segmentasi untuk setiap objek yang terdeteksi.

### 5. Segmentation Models Lain: FCN dan U-Net
Selain Mask R-CNN, bab ini juga mengenalkan beberapa model **segmentasi gambar** lainnya, seperti **Fully Convolutional Networks (FCNs)** dan **U-Net**:
*   **FCN (Fully Convolutional Network):** Model ini merupakan jenis CNN yang seluruh arsitekturnya terdiri dari layer konvolusional. Tidak ada lapisan dense dalam FCN, sehingga outputnya adalah gambar dengan ukuran yang sama dengan gambar input. FCN sering digunakan dalam segmentasi semantik (misalnya, memberi label pada objek dalam gambar).
*   **U-Net:** Model ini adalah jenis FCN yang lebih spesifik dan sangat efektif dalam segmentasi medis. U-Net menggunakan **skip connections** untuk menghubungkan lapisan encoding dan decoding, yang memungkinkan model untuk memulihkan informasi spasial yang hilang saat data diproses.

### 6. Evaluasi Kinerja Model Segmentasi
Setelah model dilatih, evaluasi kinerjanya sangat penting untuk menentukan seberapa baik model melakukan segmentasi pada data yang belum pernah dilihat sebelumnya. Beberapa metrik evaluasi yang digunakan untuk menilai kinerja model segmentasi antara lain:
*   **IoU (Intersection over Union):** Mengukur seberapa besar area tumpang tindih antara area yang diprediksi oleh model dan area yang benar-benar ada dalam gambar.
*   **Dice Coefficient:** Metrik lain yang digunakan untuk mengukur kesamaan antara dua set data. Biasanya digunakan dalam konteks segmentasi medis.
*   **Pixel Accuracy:** Mengukur persentase piksel yang benar diklasifikasikan oleh model.

### 7. Penerapan Segmentasi Gambar dalam Dunia Nyata
Bab ini juga menyajikan beberapa contoh penerapan segmentasi gambar dalam dunia nyata, termasuk:
*   **Segmentasi Medis:** Menggunakan segmentasi untuk mendeteksi **tumor** atau **lesi** dalam gambar medis seperti CT scan atau MRI.
*   **Segmentasi dalam Pengolahan Citra Satelit:** Menggunakan segmentasi untuk mengenali **tumbuhan**, **kawasan perkotaan**, dan **infrastruktur** dalam gambar satelit.

### 8. Mengoptimalkan Model Segmentasi
Untuk mengoptimalkan kinerja model segmentasi gambar, bab ini mencakup teknik-teknik tambahan seperti:
*   **Fine-tuning** model menggunakan dataset yang lebih kecil atau lebih spesifik.
*   **Transfer learning** dengan menggunakan model yang telah dilatih pada dataset besar (misalnya, ImageNet) dan menyesuaikan model tersebut untuk tugas segmentasi khusus.
