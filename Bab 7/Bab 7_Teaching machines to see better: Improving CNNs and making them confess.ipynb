# Chapter 7: Teaching machines to see better: Improving CNNs and making them confess

Bab 7 membahas teknik lanjutan untuk meningkatkan **Convolutional Neural Networks (CNNs)**, khususnya dalam meningkatkan kinerja dan efisiensi model CNN dalam tugas pengenalan gambar. Teknik-teknik ini dapat membantu model CNN menangani data yang lebih besar dan lebih kompleks serta memperbaiki hasil klasifikasi.

Berikut adalah penjelasan rinci tentang tiap bagian dari pembahasan di bab ini:

### 1. Mengatasi Overfitting dalam CNN
**Overfitting** terjadi ketika model terlalu terlatih dengan data pelatihan sehingga kehilangan kemampuannya untuk menggeneralisasi ke data baru. Bab ini membahas beberapa teknik yang dapat digunakan untuk mengatasi masalah overfitting pada model CNN:
*   **Dropout:** Salah satu teknik utama untuk mencegah overfitting adalah **dropout**, di mana neuron secara acak "dikeluarkan" selama proses pelatihan untuk mencegah model terlalu bergantung pada bagian tertentu dari data. Dropout efektif dalam meningkatkan kemampuan model untuk belajar fitur-fitur yang lebih umum dan dapat diadaptasi dengan data baru.
*   **L2 Regularization:** Teknik ini menambahkan **penalti** pada bobot model untuk mencegah bobot menjadi terlalu besar. Dengan menambahkan **L2 regularization**, model akan lebih cenderung untuk menghasilkan bobot yang lebih kecil dan lebih seimbang, yang mengurangi risiko overfitting.

### 2. Data Augmentation untuk Peningkatan Kinerja
**Data augmentation** adalah teknik untuk memperbesar dataset pelatihan dengan menciptakan variasi dari gambar yang sudah ada. Ini membantu model untuk belajar lebih banyak variasi data, sehingga meningkatkan kemampuan generalisasi dan mencegah overfitting. Beberapa teknik augmentasi yang dibahas adalah:
*   **Rotasi, flipping, dan zooming:** Mengubah orientasi gambar dengan memutarnya, membaliknya secara horizontal atau vertikal, atau mengubah tingkat pembesaran gambar untuk menghasilkan variasi.
*   **Transformasi warna dan perubahan pencahayaan:** Mengubah warna dan pencahayaan gambar untuk menciptakan kondisi pencahayaan yang berbeda dalam gambar pelatihan.
Dengan melakukan augmentasi, model dapat menjadi lebih tahan terhadap perubahan kondisi gambar dan dapat belajar fitur lebih beragam.

### 3. Menggunakan Pretrained Models untuk Transfer Learning
**Transfer learning** adalah teknik di mana model yang sudah dilatih sebelumnya pada dataset besar digunakan sebagai titik awal untuk model baru yang akan dilatih pada dataset yang lebih kecil atau spesifik. Ini memungkinkan model untuk memanfaatkan pengetahuan yang sudah ada, sehingga menghemat waktu pelatihan dan meningkatkan akurasi model.
*   **Pretrained Models:** Bab ini memperkenalkan beberapa model yang telah dilatih sebelumnya, seperti **VGG16**, **ResNet**, dan **InceptionV3**. Pembaca diajarkan bagaimana menggunakan model-model ini untuk **fine-tune** pada dataset mereka sendiri. **Fine-tuning** hanya melibatkan pelatihan ulang beberapa lapisan terakhir dari model yang sudah dilatih sebelumnya, yang mempercepat proses pelatihan dan mengurangi risiko overfitting.
*   **Contoh Penerapan:** Misalnya, model **InceptionV3** yang telah dilatih pada dataset besar seperti **ImageNet** dapat digunakan untuk klasifikasi gambar spesifik, seperti deteksi penyakit dalam gambar medis atau pengenalan objek pada dataset yang lebih kecil.

### 4. Teknik Regularisasi Lainnya
Selain **dropout** dan **L2 regularization**, bab ini juga membahas teknik-teknik lain untuk mengurangi overfitting dan meningkatkan performa model:
*   **Batch Normalization:** Batch normalization adalah teknik yang digunakan untuk **menormalkan** input ke setiap lapisan CNN. Dengan melakukan normalisasi, model menjadi lebih stabil selama pelatihan dan dapat mencapai konvergensi yang lebih cepat.
*   **Early Stopping:** Teknik ini digunakan untuk menghentikan pelatihan model lebih awal jika **akurasi validasi** tidak meningkat lagi, yang membantu menghindari overfitting. Ini memastikan bahwa model tidak dilatih terlalu lama dan menjadi terlalu spesifik pada data pelatihan.

### 5. Meningkatkan Arsitektur CNN: Dilatasi dan Sumber Daya
Di bagian ini, dibahas tentang bagaimana **mengubah arsitektur** CNN untuk menangani masalah kompleks dengan lebih baik. Modifikasi arsitektur ini dapat mencakup penggunaan **lapisan dilatasi** atau teknik-teknik lain yang memungkinkan model untuk menangani data dengan lebih banyak fitur.
*   **Dilated Convolutions:** Dalam lapisan **dilated convolution**, jarak antar filter konvolusional diperbesar, memungkinkan model untuk "melihat" lebih banyak area dalam gambar dengan menggunakan lebih sedikit parameter. Teknik ini berguna untuk tugas seperti **segmentasi gambar**, di mana kita perlu memahami konteks gambar secara lebih luas.
*   **Residual Connections:** Bab ini juga membahas **Residual Networks (ResNets)**, di mana **skip connections** ditambahkan antara lapisan-lapisan jaringan untuk mengatasi masalah **vanishing gradients** dan mempercepat pelatihan model yang sangat dalam (deep models).

### 6. Hyperparameter Tuning dan Optimisasi
Bagian ini membahas bagaimana melakukan **tuning hyperparameter** untuk meningkatkan performa model. **Hyperparameter tuning** adalah proses memilih parameter terbaik untuk model, seperti **learning rate**, **batch size**, dan **jumlah filter** pada lapisan konvolusi. Beberapa teknik untuk optimasi hyperparameter adalah:
*   **Grid Search:** Teknik di mana beberapa kombinasi hyperparameter dicoba dan evaluasi dilakukan untuk menemukan yang terbaik.
*   **Random Search:** Teknik yang lebih efisien di mana nilai-nilai hyperparameter dipilih secara acak dalam rentang tertentu, dan hasil terbaik dipilih dari kombinasi yang teruji.

### 7. Evaluasi Model dan Validasi
Setelah melatih model, evaluasi performa model adalah langkah krusial. Bagian ini membahas tentang cara melakukan **validasi silang (cross-validation)** dan **evaluasi model** untuk memastikan bahwa model tidak hanya bekerja dengan baik pada data pelatihan tetapi juga pada data yang tidak terlihat sebelumnya (data uji).
*   **Confusion Matrix:** Pembaca akan diperkenalkan dengan **confusion matrix**, yang digunakan untuk mengukur akurasi model dan membantu dalam menganalisis kesalahan klasifikasi.
