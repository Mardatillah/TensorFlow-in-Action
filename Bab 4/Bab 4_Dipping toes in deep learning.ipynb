# Chapter 4 : Dipping toes in deep learning
---
Bab 4 memberikan pengenalan langsung ke **deep learning**, fokus pada tiga jenis model dasar yang digunakan dalam banyak aplikasi pembelajaran mesin, yaitu **fully connected networks (FCNs)**, **convolutional neural networks (CNNs)**, dan **recurrent neural networks (RNNs)**. Bab ini juga mengajarkan cara mengimplementasikan dan melatih model-model dasar ini menggunakan **TensorFlow 2**.

**Berikut adalah penjelasan lebih lanjut mengenai setiap bagian dari bab ini:**

### 1. Fully Connected Networks (FCNs)
*   **Pengertian FCNs:** Fully Connected Networks adalah jenis jaringan saraf di mana setiap neuron pada satu lapisan terhubung dengan semua neuron di lapisan berikutnya. Model ini digunakan untuk tugas-tugas dasar dalam deep learning seperti klasifikasi dan regresi.
*   **Penerapan FCN:** Di bab ini, pembaca diperkenalkan pada konsep dasar **FCNs** dan cara melatih model menggunakan **TensorFlow 2**. Pembaca akan mempelajari bagaimana menyusun lapisan-lapisan dalam model FCN dan bagaimana menyusunnya secara bertahap dari input hingga output.
    *   **Contoh:** Model FCN yang sederhana bisa digunakan untuk tugas **klasifikasi** (misalnya, mengklasifikasikan gambar ke dalam kategori tertentu).

### 2. Convolutional Neural Networks (CNNs)
*   **Pengertian CNNs:** CNN adalah jenis jaringan saraf yang khusus digunakan untuk memproses data yang memiliki struktur grid seperti gambar. CNN sangat efektif dalam tugas pengenalan objek dalam gambar karena dapat mendeteksi pola dan fitur lokal menggunakan operasi konvolusi.
*   **Penerapan CNN:** Bab ini mengajarkan pembaca cara membangun model **CNN** untuk klasifikasi gambar. Dengan menggunakan operasi **konvolusi** dan **pooling**, model ini dapat mengekstraksi fitur penting dari gambar.
*   **Proses Implementasi CNN:**
    *   **Preprocessing Data:** Menyiapkan data gambar untuk digunakan dalam model CNN, seperti **resize**, **normalisasi**, dan **augmentation**.
    *   **Menggunakan Layer Konvolusional:** CNN terdiri dari beberapa lapisan konvolusional yang digunakan untuk mendeteksi fitur-fitur kecil dalam gambar (seperti tepi, sudut, tekstur, dll.).
    *   **Pelatihan dan Evaluasi:** Bab ini juga mencakup cara melatih model CNN menggunakan dataset gambar dan mengevaluasi kinerjanya.

### 3. Recurrent Neural Networks (RNNs)
*   **Pengertian RNNs:** RNN adalah jenis jaringan saraf yang sangat efektif untuk menangani data sekuensial, seperti teks atau deret waktu (time series). Berbeda dengan FCN dan CNN, RNN memiliki kemampuan untuk memori dan mengingat informasi sebelumnya dalam sekuens.
*   **Penerapan RNN:** Di bab ini, pembaca diajarkan bagaimana menggunakan **RNN** untuk memproses data sekuensial, seperti memprediksi nilai masa depan dalam data deret waktu atau menghasilkan teks. RNN dapat digunakan untuk tugas **prediksi urutan** dan **penerjemahan mesin**.
    *   **LSTM (Long Short-Term Memory)** dan **GRU (Gated Recurrent Units)** adalah dua jenis varian dari RNN yang memungkinkan jaringan untuk mengingat informasi dalam waktu yang lebih lama dan menghindari masalah **vanishing gradient**.
*   **Implementasi RNN:** Pembaca belajar cara mengimplementasikan **RNN** dalam TensorFlow, mempersiapkan data sekuensial, melatih model, dan menggunakannya untuk tugas prediksi.

### 4. Implementasi Model dan Pelatihan
*   **Memahami Data:** Sebelum melatih model, sangat penting untuk memahami data yang akan digunakan. Bab ini mengajarkan pembaca cara **mengeksplorasi data**, misalnya dengan menggunakan teknik seperti **statistik deskriptif** untuk melihat distribusi data dan pola yang ada.
*   **Proses Pelatihan Model:** Pembaca akan belajar tentang langkah-langkah pelatihan model deep learning. Ini melibatkan **optimasi**, **penentuan loss function**, dan **evaluasi** model menggunakan data uji (test data).
    *   **Backpropagation** digunakan untuk memperbarui bobot model selama pelatihan, sementara **optimizer** seperti **Adam** dan **SGD** digunakan untuk mempercepat proses pelatihan.

### 5. Penerapan Model pada Masalah Dunia Nyata
*   **Contoh Penggunaan FCNs, CNNs, dan RNNs dalam Dunia Nyata:** Di akhir bab, dibahas aplikasi nyata dari model-model yang telah dipelajari, seperti:
    *   **FCNs** untuk klasifikasi dasar, misalnya, klasifikasi gambar atau teks.
    *   **CNNs** untuk tugas pengenalan gambar atau deteksi objek.
    *   **RNNs** untuk aplikasi **pemrosesan bahasa alami (NLP)**, seperti penerjemahan atau analisis sentimen.
