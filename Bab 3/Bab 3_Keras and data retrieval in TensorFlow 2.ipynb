1. Keras Model-Building APIs
    
    Keras adalah API tingkat tinggi yang sangat berguna untuk membangun model pembelajaran mesin, terutama untuk deep learning. Keras menyediakan berbagai **model-building API** untuk membuat jaringan saraf secara lebih mudah dan cepat. Berikut penjelasan tentang masing-masing API yang dibahas di bab ini:

    *   **API Sequential:** API ini digunakan untuk membangun model dengan urutan layer yang sederhana. Model ini cocok untuk jaringan saraf yang terdiri dari urutan linear layer, seperti dalam model **fully connected networks**.

        *   **Contoh:** Model Sequential akan memiliki input layer, beberapa hidden layer, dan output layer yang saling terhubung secara linier.

    *   **API Functional:** API ini memungkinkan pembuat model untuk membuat jaringan saraf yang lebih kompleks, di mana koneksi antar layer tidak harus berbentuk linear. Model ini cocok untuk multi-input/output atau skip connections.

        *   **Contoh:** Model yang membutuhkan layer dengan input/output yang berbeda atau ketika kita ingin membuat ResNet (Residual Network).

    *   **API Sub-classing:** API ini memungkinkan pengguna untuk membuat model dengan lebih banyak fleksibilitas dengan cara membuat kelas custom. Ini memberi kontrol lebih tinggi, memungkinkan pembuatan layer atau model yang lebih kompleks dan unik. Subclassing cocok untuk model yang sangat kompleks atau eksperimen.

        *   **Contoh:** Model yang menggunakan reinforcement learning atau algoritma eksperimen lainnya yang membutuhkan kontrol lebih besar.

2. Mengambil Data untuk Model TensorFlow/Keras

    Bab ini juga mengulas cara-cara mengimpor dan memproses data untuk digunakan dalam pelatihan model di TensorFlow. Pengambilan data adalah salah satu langkah penting dalam siklus pelatihan model deep learning. Beberapa metode yang dibahas untuk mengambil dan mempersiapkan data adalah:

    *   **tf.data API:** Merupakan API utama untuk membuat pipeline data yang efisien di TensorFlow. Dengan API ini, kita bisa membuat pipeline untuk mengambil data dalam bentuk yang lebih terstruktur dan memprosesnya sebelum dimasukkan ke model.

        *   **Contoh:** Menyiapkan dataset gambar dan memanipulasinya (resize, normalisasi) sebelum digunakan untuk pelatihan model.

    *   **Keras DataGenerators:** Data generators adalah metode yang berguna untuk mengambil dan memproses data secara **batch** dalam pembelajaran mesin. Dengan menggunakan DataGenerators, kita bisa mengimpor data gambar atau teks dan memberikan data ke model dalam **batch** yang lebih kecil. Ini membantu model untuk memproses data besar yang tidak muat dalam memori sekaligus.

        *   **Contoh:** Menggunakan **ImageDataGenerator** untuk memuat data gambar dari folder dan meng-aplikasikan augmentasi (misalnya, rotasi, flipping) untuk meningkatkan variasi data pelatihan.

    *   **tensorflow-datasets package:** TensorFlow menyediakan dataset siap pakai melalui tensorflow-datasets, yang merupakan kumpulan dataset populer yang sudah terorganisir dan siap digunakan dalam model pembelajaran mesin.

        *   **Contoh:** Mengambil dataset seperti **CIFAR-10** atau **MNIST** yang digunakan untuk tugas klasifikasi gambar.

3. Preprocessing dan Augmentasi Data

    Di bagian ini, dibahas tentang cara memproses dan meningkatkan data agar model dapat dilatih lebih baik. Salah satu teknik penting yang diperkenalkan adalah **data augmentation**, yang membantu menghindari **overfitting** dan meningkatkan generalisasi model. Beberapa teknik augmentasi yang umum adalah rotasi gambar, flipping, zooming, dan perubahan warna.

    *   **Augmentasi Gambar:** Teknik ini mengubah data pelatihan (misalnya, gambar) untuk menciptakan variasi yang lebih besar. Dengan augmentasi gambar, kita dapat mengubah gambar sedikit (misalnya, rotasi atau flipping) untuk membantu model belajar lebih banyak variasi dari gambar yang sama.

    *   **Preprocessing Data:** Selain augmentasi, preprocessing juga penting untuk mempersiapkan data sebelum dimasukkan ke dalam model. Ini melibatkan normalisasi, pengubahan ukuran, dan konversi data ke format yang sesuai untuk model.

        *   **Contoh:** Mengubah ukuran gambar ke dimensi yang sesuai (misalnya, 224x224 piksel) dan menormalkan nilai piksel agar berada dalam rentang 0 hingga 1.
