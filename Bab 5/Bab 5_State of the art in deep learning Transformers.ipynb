# Chapter 5 : State-of-the-art in deep learning: Transformers

Bab 5 membahas **Transformer models**, yang telah menjadi paradigma baru dalam **deep learning**, khususnya dalam tugas-tugas **pemrosesan bahasa alami (NLP)**. Transformer memungkinkan pemrosesan data sekuensial secara lebih efisien dan dapat melihat keseluruhan urutan data sekaligus, berbeda dengan model sebelumnya seperti **RNN** yang hanya dapat melihat bagian data pada satu waktu. Model ini sangat populer di dunia NLP, terutama dalam aplikasi seperti **penerjemahan mesin** dan **generasi teks**.

**Berikut adalah penjelasan rinci dari setiap bagian yang dibahas di bab ini:**

### 1. Representasi Teks sebagai Angka
Bagian pertama dari bab ini membahas cara representasi teks dalam bentuk **numerik** yang dapat diproses oleh model deep learning. Teks, yang merupakan data non-numerik, harus diubah menjadi angka sebelum dimasukkan ke dalam model deep learning. Proses ini sering melibatkan:
*   **Tokenisasi:** Memecah teks menjadi unit-unit lebih kecil (token), seperti kata atau karakter.
*   **Penyandian:** Mengonversi token-token tersebut menjadi angka, sering kali menggunakan **one-hot encoding** atau **word embeddings** seperti **Word2Vec** atau **GloVe**.

Teks yang telah diubah menjadi representasi numerik ini dapat diproses lebih lanjut oleh model deep learning, termasuk Transformer.

### 2. Memahami Model Transformer
Bab ini memberikan pengenalan mendalam tentang arsitektur **Transformer**, yang pertama kali diperkenalkan dalam makalah "Attention is All You Need" oleh Vaswani et al. (2017). Model Transformer secara drastis mengubah cara kita memproses urutan data dibandingkan dengan pendekatan sebelumnya, seperti **RNN** dan **LSTM**. Berikut adalah elemen utama yang membentuk Transformer:
*   **Self-Attention Mechanism:** Komponen inti dari Transformer yang memungkinkan model untuk memberi perhatian pada semua bagian dari urutan input secara simultan, bukan satu bagian pada satu waktu. Ini memungkinkan model untuk memahami konteks yang lebih luas dalam data sekuensial.
*   **Multi-Head Attention:** Dalam multi-head attention, ada beberapa "kepala" yang memungkinkan model untuk mempelajari berbagai representasi dari informasi yang sama secara paralel.
*   **Feed-forward Layers:** Setelah melalui lapisan perhatian, data diproses lebih lanjut oleh lapisan feed-forward (lapisan yang diterapkan pada setiap posisi secara independen).
*   **Positional Encoding:** Karena Transformer tidak memproses data secara urut, informasi urutan diberikan dengan menggunakan **positional encoding** yang menambahkan informasi tentang posisi relatif dari elemen-elemen dalam urutan.

### 3. Self-Attention: Mengerti Konsep
Bagian ini membahas lebih dalam tentang **self-attention**, yang merupakan mekanisme kunci dalam model Transformer. Self-attention memungkinkan model untuk memfokuskan perhatian pada bagian data yang relevan di seluruh urutan, memperhitungkan hubungan antar elemen dalam sekuens, tanpa harus mengandalkan pemrosesan urutan langkah demi langkah seperti yang terjadi di **RNN**.
*   **Self-attention** bekerja dengan cara menghitung tiga vektor untuk setiap kata (atau token) dalam urutan: **Query**, **Key**, dan **Value**. Melalui operasi perhitungan antara query dan key, model dapat menentukan seberapa besar perhatian yang diberikan pada setiap kata dalam urutan tersebut.

### 4. Memahami Self-Attention Menggunakan Skalar
Untuk memberikan pemahaman yang lebih baik, bab ini menyajikan contoh bagaimana self-attention bekerja menggunakan **skalar**. Misalnya, ketika suatu kata dalam kalimat ingin "memahami" kata lain yang lebih jauh dalam urutan, self-attention mengukur seberapa banyak perhatian yang perlu diberikan ke kata tersebut, sehingga menghasilkan representasi yang lebih kaya dan informatif.

### 5. Masked Self-Attention Layers
Selain perhatian biasa, bab ini juga membahas tentang **masked self-attention** yang digunakan dalam tugas seperti **penerjemahan**. Di sini, Transformer tidak melihat masa depan dalam urutan selama pelatihan, yang penting agar model tidak mengakses informasi yang seharusnya belum tersedia.
*   **Masked attention** membatasi perhatian hanya pada kata-kata yang sudah ada dalam urutan dan mencegah model melihat informasi masa depan yang belum diproses.

### 6. Multi-Head Attention
Transformer menggunakan **multi-head attention**, yang memperkenalkan beberapa kepala perhatian yang berjalan paralel. Setiap kepala dapat belajar representasi yang berbeda dari urutan yang sama, yang memungkinkan model untuk mempelajari berbagai fitur atau aspek informasi dalam data secara bersamaan.
*   Multi-head attention mengoptimalkan cara model menangkap berbagai dependensi dalam data, meningkatkan kemampuannya untuk memproses urutan yang lebih panjang dengan lebih baik.

### 7. Fully Connected Layer
Setelah perhatian selesai, data yang telah diproses oleh lapisan perhatian dimasukkan ke dalam **fully connected layer** (lapisan sepenuhnya terhubung). Lapisan ini memungkinkan model untuk memproses informasi lebih lanjut dan menghasilkan output yang lebih kompleks, seperti prediksi kelas dalam tugas klasifikasi atau token dalam penerjemahan.

### 8. Menyatukan Semuanya
Setelah membahas mekanisme dasar, bab ini menjelaskan bagaimana semua komponen tersebut digabungkan menjadi satu model yang koheren. Model Transformer terdiri dari beberapa **encoder** dan **decoder** yang bekerja secara bersamaan untuk menangani data input dan output.
*   **Encoder:** Mengambil input dan memprosesnya melalui lapisan self-attention untuk menghasilkan representasi yang lebih kaya.
*   **Decoder:** Mengambil representasi dari encoder dan memprosesnya untuk menghasilkan output (misalnya, dalam penerjemahan).
