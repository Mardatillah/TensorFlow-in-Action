# Chapter 6 : Teaching machines to see: Image classification with CNNs
---
Bab 6 membahas tentang penerapan **Convolutional Neural Networks (CNNs)** untuk **klasifikasi gambar**. CNN adalah jenis jaringan saraf yang paling sering digunakan untuk memproses data gambar dan video, karena kemampuannya untuk mengenali pola atau fitur dalam data gambar. Bab ini fokus pada cara mengimplementasikan dan melatih CNN menggunakan **TensorFlow 2** dan **Keras**, serta cara menangani dan mempersiapkan data gambar untuk pelatihan model.

Berikut adalah penjelasan rinci dari setiap bagian pembahasan dalam bab ini:

### 1. Memasukkan Data ke dalam Model: Eksplorasi Data
Sebelum melatih model CNN, sangat penting untuk memahami data yang akan digunakan. Bab ini mengajarkan bagaimana cara melakukan **eksplorasi data** untuk gambar:
*   **Struktur Folder dan File:** Gambar biasanya disimpan dalam folder berdasarkan kategori atau label, seperti folder untuk gambar **anjing** dan folder untuk gambar **kucing**. Pembaca akan belajar bagaimana mempersiapkan data gambar ini untuk digunakan dalam model.
*   **Statistik Sederhana:** Menghitung statistik sederhana tentang dataset, seperti jumlah gambar per kelas, ukuran gambar, dan distribusi data. Ini membantu dalam memahami data yang akan digunakan dan mengevaluasi apakah dataset tersebut seimbang atau tidak.

### 2. Membuat Pipeline Data Menggunakan Keras ImageDataGenerator
Untuk menangani dataset gambar dalam skala besar, kita membutuhkan pipeline data yang efisien. **Keras ImageDataGenerator** adalah alat yang sangat berguna untuk memuat gambar dari folder dan mengaplikasikan transformasi secara real-time saat model sedang dilatih.
*   **Pembuatan Pipeline:** Pengguna akan diajarkan cara menggunakan **ImageDataGenerator** untuk membaca gambar dari disk, melakukan **augmentation** gambar (misalnya, rotasi, flipping, zooming), dan menormalisasi gambar sebelum dimasukkan ke model.
*   **Data Augmentation:** Teknik ini digunakan untuk memperbesar variasi data pelatihan dengan memodifikasi gambar asli, sehingga model dapat belajar lebih banyak fitur dan mengurangi risiko overfitting.

### 3. Inception Net: Mengimplementasikan Klasifikasi Gambar dengan Model CNN Canggih
Setelah memahami dasar-dasar CNN, bab ini melanjutkan dengan menggunakan **Inception Net**, yang merupakan salah satu model CNN yang lebih canggih dan digunakan secara luas untuk klasifikasi gambar.
*   **Inception V1:** Pembaca diajarkan bagaimana mengimplementasikan **Inception v1** menggunakan **Keras**. Inception adalah arsitektur CNN yang dirancang untuk memanfaatkan berbagai ukuran kernel konvolusi dalam satu lapisan, yang memungkinkan model untuk menangkap berbagai fitur dari gambar.
*   **Menggabungkan Komponen:** Bagian ini mengajarkan bagaimana menyusun komponen-komponen Inception seperti **1x1 convolutions**, **3x3 convolutions**, dan **5x5 convolutions** dalam sebuah blok dan menggabungkannya untuk meningkatkan kemampuan model dalam mengenali pola-pola yang lebih kompleks dalam gambar.

### 4. Melatih Model dan Mengevaluasi Kinerja
Setelah model CNN (termasuk Inception) dibangun, langkah berikutnya adalah melatih model menggunakan dataset gambar. Bab ini membahas berbagai langkah yang terlibat dalam melatih dan mengevaluasi model:
*   **Training:** Melatih model dengan menggunakan dataset yang sudah dipersiapkan dan mengoptimalkan model untuk meminimalkan **loss function** (fungsi kerugian) menggunakan algoritma optimasi seperti **Adam**.
*   **Evaluasi Model:** Setelah melatih model, langkah berikutnya adalah mengevaluasi kinerjanya. **Akurasi** adalah metrik utama yang digunakan untuk mengukur kinerja model klasifikasi. Bab ini juga membahas cara mengukur **precision**, **recall**, dan **F1-score** untuk mengevaluasi model lebih lanjut, terutama pada dataset yang tidak seimbang.

### 5. Mengoptimalkan Model dan Menghindari Overfitting
Pengoptimalan model adalah langkah penting dalam meningkatkan akurasi dan mengurangi **overfitting**, di mana model belajar terlalu banyak dari data pelatihan dan gagal menggeneralisasi ke data baru.
*   **Regularisasi:** Teknik seperti **dropout** dan **L2 regularization** dibahas untuk membantu mengurangi overfitting. Dropout mengacak-nekan beberapa neuron selama pelatihan untuk mencegah model terlalu bergantung pada fitur tertentu.
*   **Early Stopping:** Penggunaan teknik **early stopping** untuk menghentikan pelatihan lebih awal jika model mulai menunjukkan tanda-tanda overfitting.

### 6. Transfer Learning: Menggunakan Jaringan yang Sudah Dilatih
Bab ini juga memperkenalkan konsep **transfer learning**, yang memungkinkan kita untuk memanfaatkan model CNN yang sudah dilatih sebelumnya untuk mempercepat proses pelatihan pada dataset kita.
*   **Pretrained Models:** Pembaca akan belajar bagaimana menggunakan model yang sudah dilatih sebelumnya, seperti **InceptionV3** atau **VGG16**, dan hanya melatih beberapa lapisan akhir untuk menyelesaikan tugas klasifikasi pada dataset baru. Ini memungkinkan kita untuk memanfaatkan pengetahuan yang sudah diperoleh dari dataset besar dan mempercepat pelatihan pada dataset kecil.
