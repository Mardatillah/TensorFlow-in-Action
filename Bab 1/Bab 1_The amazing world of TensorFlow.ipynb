1. Apa Itu TensorFlow?

    Pada bagian ini, pembaca dikenalkan dengan **TensorFlow** sebagai framework pembelajaran mesin (machine learning) yang sangat penting dan banyak digunakan. TensorFlow adalah **framework end-to-end**, yang berarti dapat digunakan untuk seluruh siklus hidup model pembelajaran mesin, mulai dari pembangunan, pelatihan, hingga penerapan ke dunia nyata. TensorFlow membantu **memroses data** besar (seperti gambar atau teks) dan membuatnya lebih berguna untuk analisis dan prediksi. TensorFlow memiliki kemampuan untuk menjalankan perhitungan di **hardware yang dioptimalkan**, seperti **GPU** (Graphics Processing Unit) dan **TPU** (Tensor Processing Unit).

2. Perangkat Keras dalam Pembelajaran Mesin: GPU vs. CPU

    Di bagian ini, dibahas perbedaan antara **CPU** (Central Processing Unit) dan **GPU** yang digunakan dalam pembelajaran mesin. **GPU** lebih efisien untuk melakukan operasi paralel yang dibutuhkan oleh model deep learning, sementara **CPU** lebih cocok untuk tugas yang lebih kompleks tetapi pada skala yang lebih kecil. **TPU**, yang merupakan perangkat keras khusus yang dikembangkan oleh Google, juga diperkenalkan sebagai alat yang dioptimalkan untuk mempercepat perhitungan dalam pembelajaran mesin.

3. Kapan Harus Menggunakan TensorFlow
    
    TensorFlow sangat efektif untuk **prototyping model deep learning**, seperti **fully connected networks**, **convolutional neural networks (CNNs)**, dan **recurrent neural networks (RNNs)**. TensorFlow sangat berguna saat Anda ingin **mempercepat pelatihan model** dengan memanfaatkan perangkat keras yang dioptimalkan (seperti GPU atau TPU). Framework ini juga cocok untuk **menyajikan model ke publik** melalui API atau layanan cloud setelah model selesai dilatih.
    *   **Prototyping Deep Learning Models:** TensorFlow memungkinkan pembuatan model deep learning dengan menggunakan layer objek (seperti Dense layers untuk fully connected networks, Convolution layers untuk CNNs, dan LSTM layers untuk RNNs).
    *   **Implementing Models Faster on Optimized Hardware:** TensorFlow dapat mempercepat perhitungan dengan memanfaatkan hardware yang dioptimalkan (seperti GPU atau TPU) untuk mempercepat pelatihan dan inferensi model.

4. Kapan Tidak Menggunakan TensorFlow

    Meskipun TensorFlow sangat berguna, ada kalanya penggunaan TensorFlow bukan pilihan terbaik. Untuk tugas **machine learning tradisional** seperti **regresi linear/logistik**, **decision trees**, atau **k-means**, pustaka lain seperti **scikit-learn** lebih efisien. TensorFlow lebih cocok untuk **deep learning** dan tidak ideal untuk menangani **data kecil** yang bisa dengan mudah dimanipulasi menggunakan **pandas** atau **NumPy**.
    *   **Implementing Traditional Machine Learning Models:** Untuk model tradisional yang tidak memerlukan komputasi paralel, scikit-learn adalah pilihan yang lebih sederhana dan efisien.
    *   **Manipulating and Analyzing Small-Scale Structured Data:** Jika bekerja dengan dataset kecil yang bisa dimuat sepenuhnya dalam memori, menggunakan **pandas** dan **NumPy** lebih efisien daripada TensorFlow.
